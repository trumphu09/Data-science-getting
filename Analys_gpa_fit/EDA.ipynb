```python
df1 = pd.read_csv("K70IT1(Sheet1).csv", encoding='latin1', skiprows=3)
df2 = pd.read_csv("K70IT2(Sheet1).csv", encoding='latin1', skiprows=3)
df3 = pd.read_csv("K70IT3(Sheet1).csv", encoding='latin1', skiprows=3)  
df4 = pd.read_csv("K70IT4(Sheet1).csv", encoding='latin1', skiprows=3)
df5 = pd.read_csv("K70IT5(Sheet1).csv", encoding='latin1', skiprows=3)
df6 = pd.read_csv("K70IT6(Sheet1).csv", encoding='latin1', skiprows=3)
df7 = pd.read_csv("K70IT7(Sheet1).csv", encoding='latin1', skiprows=3)
```

```python
dfIT = pd.concat([ df1, df2, df3, df4, df5, df6, df7], ignore_index=True)
```

```python
dfIT.describe()
```

```python
dfIT = dfIT.sort_values(by = "Unnamed: 6", ascending=False)
```

```python
# Kiểm tra dữ liệu gốc dfIT
print("Kích thước dfIT:", dfIT.shape)
print("\nDòng đầu tiên:")
print(dfIT.head(3))
print("\nKiểu dữ liệu cột Unnamed: 6:")
print(dfIT['Unnamed: 6'].dtype)
print("\nSample giá trị cột Unnamed: 6:")
print(dfIT['Unnamed: 6'].head(20))
```

```python
# Tạo copy của dfIT để xử lý
dfIT_group = dfIT.copy()

# Bước 1: Xóa ký tự \xa0 (non-breaking space) và strip whitespace
dfIT_group['Unnamed: 6'] = dfIT_group['Unnamed: 6'].str.replace('\xa0', '').str.strip()

# Bước 2: Chuyển đổi sang numeric (xóa string không hợp lệ)
dfIT_group['Unnamed: 6'] = pd.to_numeric(dfIT_group['Unnamed: 6'], errors='coerce')

# Bước 3: Xóa các hàng có NaN ở cột điểm
dfIT_group = dfIT_group.dropna(subset=['Unnamed: 6'])

print(f"✓ Chuyển đổi thành công!")
print(f"Số hàng dữ liệu: {len(dfIT_group)}")
print(f"Điểm min: {dfIT_group['Unnamed: 6'].min()}")
print(f"Điểm max: {dfIT_group['Unnamed: 6'].max()}")
print(f"\nSample dữ liệu:")
print(dfIT_group[['Unnamed: 2', 'Unnamed: 6']].head())
```

```python
# Kiểm tra chi tiết dữ liệu cột Unnamed: 6
print("Giá trị unique:")
print(dfIT['Unnamed: 6'].unique()[:20])
print("\nSố giá trị null:", dfIT['Unnamed: 6'].isna().sum())
print("\nCó giá trị 'NaN' (string)?", ('NaN' in dfIT['Unnamed: 6'].values))
print("\nSample 20 giá trị:")
for i, val in enumerate(dfIT['Unnamed: 6'].head(20).values):
    print(f"  [{i}] repr={repr(val)}, type={type(val).__name__}")
```

```python
# Tạo cột khoảng điểm bằng pd.cut() (sạch sẽ hơn)
max_score = dfIT_group['Unnamed: 6'].max()
dfIT_group['score_range'] = pd.cut(
    dfIT_group['Unnamed: 6'],
    bins=np.arange(0, int(max_score) + 1, 0.5),
    right=False
)

# Tổng hợp số lượng theo khoảng điểm
dfIT_group_summary = dfIT_group.groupby('score_range', observed=True).size().reset_index(name='count')
dfIT_group_summary
```
